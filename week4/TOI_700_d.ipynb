```python
#!pip install -q lightkurve
import lightkurve as lk
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1. Search for TESS light curves of TOI-700
search_results = lk.search_lightcurve("TOI-700", mission="TESS", author="SPOC")
print(f"Total number of sectors found: {len(search_results)}")
```

```python
# 2. Download the first available sector and clean noise
lc = search_results[0].download()
lc_flat = lc.select_flux('pdcsap_flux').remove_nans().flatten()
```

```python
# 3. Plot the Light Curve
fig, ax = plt.subplots(figsize=(10, 5))
lc_flat.scatter(ax=ax, color='black', alpha=0.6, s=2)
plt.title("TESS Light Curve of TOI-700 (Sector 1 - PDCSAP Flux)", fontsize=12)
plt.xlabel("Time - 2457000 [BTJD days]")
plt.ylabel("Normalized Flux")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

### Observation Analysis of TOI-700 Light Curve

Upon processing the TESS Sector 1 photometric data for TOI-700 (TIC 150428135) using the `lightkurve` package, I plotted the normalized Pre-search Data Conditioning Simple Aperture Photometry (`pdcsap_flux`). The time-series light curve displays an exceptionally stable continuum centered at 1.000, with a low photometric scatter of approximately ±0.4%. This quiet baseline confirms that TOI-700 is a magnetically inactive M-dwarf star, exhibiting negligible stellar flaring or large spot modulation. A prominent observational data gap is visible around BTJD 1338–1340, corresponding to the standard TESS spacecraft orbit perigee for data downlink.

I selected TOI-700 because it hosts TOI-700 d, the first Earth-sized habitable-zone exoplanet discovered by TESS. In this raw, unbinned single-sector plot, individual planetary transits are not immediately discernible to the naked eye because the transit depth of TOI-700 d (~0.1%) is shallow and embedded beneath the ~0.4% photometric scatter. This practical result perfectly illustrates why raw time-series data alone is insufficient for detecting terrestrial planets, highlighting the absolute necessity of applying Box-Least-Squares (BLS) periodograms, phase-folding, and binning to extract faint exoplanetary signals from space photometry.

### Exoplanet Discovery Report: TOI-700 d

---

#### 1. Summary Information
* **Name of the Planet:** TOI-700 d
* **Host Star:** TOI-700 (TIC 150428135) — An M2V type red dwarf star located approximately 101.5 light-years away in the constellation Dorado.
* **Planet Radius:** 1.19 ± 0.11 R_Earth (approximately 1.19 Earth radii).
* **Orbital Period:** 37.426 days.
* **Detection Method:** Transit Photometry (Discovered by NASA's TESS Mission and later validated with the Spitzer Space Telescope).

---

#### 2. Detailed Overview & System Characteristics

* **System Context & Orbital Dynamics:**
  TOI-700 d is the outermost known exoplanet in the multi-planetary TOI-700 system, which also hosts TOI-700 b (a rocky planet orbiting in 9.98 days), TOI-700 c (a sub-Neptune orbiting in 16.05 days), and TOI-700 e. TOI-700 d orbits its host star at a distance of roughly 0.163 AU. Despite being much closer to its star than Mercury is to the Sun, the low luminosity of the M-dwarf host star places TOI-700 d squarely inside the conservative Habitable Zone, receiving about 86% of the solar flux that Earth receives from the Sun.

* **Interesting Fact About the System:**
  Unlike most active M-dwarf stars (such as Proxima Centauri or TRAPPIST-1), which produce frequent intense UV flares and strong stellar winds that can strip planetary atmospheres, TOI-700 is extraordinarily quiet. Over multiple TESS sectors, no major flaring events were detected, making TOI-700 d one of the most promising targets for atmospheric retention and habitability studies.

* **Motivation for Selection:**
  I selected TOI-700 d because it represents a major milestone for NASA's TESS mission as its first validated terrestrial-sized, habitable-zone exoplanet. Studying TOI-700 d provides a perfect bridge between transit photometry discovery methods and future atmospheric characterization using transmission spectroscopy with space observatories such as the James Webb Space Telescope (JWST).
